# Ejemplo del área de un circulo

Un experimento determinista es aquel que produce siempre el mismo resultado cuando se repite bajo las mismas condiciones iniciales.

```python
import sympy

def area(radio):
    return sympy.pi * (radio ** 2)
```

# Experimento aleatorio

Lógica Lanzamiento de un dado

```python
import random

dados_lados = ["1", "2", "3", "4", "5", "6"]

def lanzar_dados():
    return random.choice(dados_lados)
```

```python
lanzar_dados()
```

Lanzamiento de una moneda

```python
import random

moneda_lados = ["aguila", "sol"]

def lanzar_moneda():
    return random.choice(moneda_lados)
```

```python
lanzar_moneda()
```
